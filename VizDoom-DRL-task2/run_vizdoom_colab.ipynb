```python
# Display training results
import json
import pandas as pd

# Load results
with open('/content/gdrive/MyDrive/ML4_vizdoom/results.json', 'r') as f:
    results = json.load(f)

# Convert to DataFrame for better visualization
df = pd.DataFrame()
for policy_type, policy_results in results.items():
    for seed, result in enumerate(policy_results):
        result['policy'] = policy_type
        result['seed'] = seed
        df = pd.concat([df, pd.DataFrame([result])], ignore_index=True)

# Display results
print("Training Results:")
print(df.groupby('policy').agg({
    'mean_reward': ['mean', 'std'],
    'mean_episode_length': ['mean', 'std'],
    'train_time': ['mean', 'std']
}).round(2))
```
